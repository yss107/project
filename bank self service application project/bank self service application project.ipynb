```python
from tabulate import tabulate
import random
import datetime


account_details = {
    "full_name": "",
    "initial_deposit": 0,
    "account_number": str(random.randint(100000000000, 999999999999))
} 


transaction_history = []  
transaction_id = 125478665412  

account_details["full_name"] = input("Please enter your full name: ")
account_details["initial_deposit"] = float(input("Please enter your initial deposit amount: "))

def perform_transaction(transaction_type): 
    
    global    transaction_id 
    if transaction_type == 'Debit': 
        transaction_amount = float(input("Enter transaction amount: "))
        if transaction_amount > account_details["initial_deposit"]:
            print("Insufficient balance for debit transaction.")
        else:
            account_details["initial_deposit"] -= transaction_amount
            transaction_id = transaction_id+1
            transaction_history.append((datetime.datetime.now(), transaction_id, transaction_type, transaction_amount, account_details["initial_deposit"],"OK"))
            print("Debit transaction successful.")
    elif transaction_type == 'Credit':
        transaction_amount = float(input("Enter transaction amount: "))
        account_details["initial_deposit"] += transaction_amount
        transaction_id = transaction_id+1 
        transaction_history.append((datetime.datetime.now(), transaction_id, transaction_type, transaction_amount, account_details["initial_deposit"],"OK"))
        print("Credit transaction successful.")
    elif transaction_type == 'Print statement': 
         print("Account Holder Name:", account_details["full_name"])
         print("Account Number: ",str(account_details[ "account_number"])[-4:].rjust(12,'*'),sep="")

         print("Bank Name: State Bank of India")
         print("City: Jodhpur")
         print(tabulate(transaction_history, headers=['Transaction Date', 'Transaction ID', 'Transaction Type', 'Transaction Amount', 'Available Balance',"Remark"], tablefmt='fancy_grid'))
    elif transaction_type == 'Check Balance':
        print("Available balance:", account_details["initial_deposit"])

while True:
    print("Select a transaction type:")
    print("1. Debit")
    print("2. Credit")
    print("3. Print statement")
    print("4. Check Balance")
    print("5. Exit")
    
    choice = input("Enter your choice: ")
    
    if choice == '1':
        perform_transaction('Debit')
    elif choice == '2':
        perform_transaction('Credit')
    elif choice == '3':
        perform_transaction('Print statement')
    elif choice == '4':
        perform_transaction('Check Balance')
    elif choice == '5':
        break
    else:
        print("Invalid choice. Please select a valid option.")
        


```
