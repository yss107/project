### Personal Expense Tracker

To user, you need to give option to
 1. Add Expense
 2. Delete an Expense
 3. Edit (update) any expense
 4. View Summary of Expenses and 
 5. FIlter Summary by Month, Year or Category
    

While Adding expnese, we need to capture:
1. Expense Category
2. Expense Sub-Category
3. Expense Description
4. Quantity
5. Total Amount Spent
6. Expense Date
7. Expense Mode: UPI, Cash, Credit Card, Debit Card
8. Seller
