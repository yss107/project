## 2. Loading the Dataset <a id='loading'></a>

Load the SYNTH dataset from HuggingFace using our custom loader.

```python
# Initialize the loader
loader = SYNTHLoader()

# Load the dataset
print("Loading dataset from HuggingFace...")
dataset = loader.load_dataset()

print("\n✓ Dataset loaded successfully!")
```

```python
# Get dataset information
info = loader.get_dataset_info()

print("Dataset Information:")
print(f"  Name: {info['name']}")
print(f"  Splits: {info['splits']}")
print(f"  Number of rows: {info['num_rows']}")
print(f"\nFeatures: {list(info['features'].keys())}")
```

```python
# Convert to pandas DataFrame for easier manipulation
df = loader.load_as_pandas(split='train', max_rows=10000)

print(f"✓ Loaded {len(df)} rows into DataFrame")
print(f"\nDataFrame shape: {df.shape}")
df.head()
```
